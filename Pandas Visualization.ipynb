```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("College_Admission.csv")
```

```python

Admission_rate = (df['admission_status'].value_counts())
Admission_rate
```

```python
# Stream-wise admissions

Stream_admission = df.groupby('preferred_stream')['admission_status'].value_counts()
Stream_admission
```

```python
# Top 10 students based on boards percentage

Top_students = df.sort_values(by='board_percentage', ascending=False).head(10)
Top_students
```

```python
# Gender-wise Admissions

Gender_admission = (df.groupby('gender')['admission_status'].value_counts())
Gender_admission
```

```python
# Entrance exam cut-off 

df.groupby('admission_status')['entrance_score'].mean()
```

```python
# Category-wise Admission

Category_analysis = (df['category'], df['admission_status'])
Category_analysis
```

```python
df['category'].value_counts().plot(kind='bar', stacked=True)
plt.title('Category-wise Admission')
plt.show()
```

```python
df['admission_status'].value_counts().plot(kind='bar')
plt.title('admission_status')
plt.show()
```

```python
plt.scatter(df['board_percentage'], df['entrance_score'])
plt.xlabel('board_percentage')
plt.ylabel('entrance_score')
plt.title('Percentege vs Entrance Score')
plt.show()
```

```python
sns.boxplot(x='admission_status', y='entrance_score', data=df)
plt.show()
```

```python
# Most popular stream

df['preferred_stream'].value_counts().plot(kind='bar')
plt.title('Stream Demand')
plt.show()
```
